# Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/DSA_2050_Week1_Retail_Data.csv')
```

```python
df.head()
#each row in the dataset represents a single transaction, with details about the product, quantity, price, and customer information.
```

```python
df.shape
#there are 300 rows and 11 columns in the dataset
```

```python
df.info
#the main data types in the dataset are integers and strings. The dataset contains no missing values.
```

## KPIs

```python
# total revenue 
Total_Revenue = df['Revenue_KES'].sum()
print(f'Total Revenue: KSH {Total_Revenue} ')    
# the total revenue generated from all transactions in the dataset.
```

```python
#Number of unique customers 
Unique_Customers = df['CustomerID'].nunique()
print(f'Unique Customers: {Unique_Customers}')  
# the number of unique customers who made purchases in the dataset.
```

```python
#average revenue per transaction
Average_Revenue_per_Transaction = Total_Revenue / df.shape[0]
print(f'Average Revenue per Transaction: KSH {Average_Revenue_per_Transaction   }')
# the average revenue generated per transaction in the dataset.
```

## Visualisations

```python
#total revenue by product
Revenue_by_Product = df.groupby('Product')['Revenue_KES'].sum()
Revenue_by_Product.plot(kind='bar', title='Total Revenue generated by Each Product', ylabel='Revenue (KES)', xlabel='Product')
plt.grid(axis='y', linestyle='--', alpha=0.7)  # horizontal grid lines
plt.show()
# A bar chart showing the total revenue generated by each product in the dataset.
# Office Chairs generated the highest revenue, while charging cables generated the lowest revenue.
```

```python
#total revenue by Region
Revenue_by_Region = df.groupby('Region')['Revenue_KES'].sum()
Revenue_by_Region.plot(kind='bar', title='Total Revenue generated from Each Region', ylabel='Revenue (KES)', xlabel='Region' )
plt.ylim(700000, 900000)
plt.grid(axis='y', linestyle='--', alpha=0.7) 
plt.show()
# A bar chart showing the total revenue generated by each region in the dataset.
# The kisumu region generated the highest revenue, while the Nairobi region generated the lowest revenue
```

####     In Conclusion, the analysis of the retail dataset provides valuable insights into the performance of different product categories and regions. The total revenue generated from all transactions is KSH 3,206,450 , with an average revenue per transaction of KSH 10,688.167. The dataset contains 300 transactions and 11 columns, with no missing values. The analysis also reveals that Office Chairs generated the highest revenue among product categories, while charging cables generated the lowest revenue. Additionally, the Kisumu region generated the highest revenue among regions, while the Nairobi region generated the lowest revenue. These insights help us understand the overall business performance and where this performance is concentrated.
